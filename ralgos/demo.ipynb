## Import

```python
# generator: generators
from ralgos.generator import uniform, uuniform, kuniform, bernoulli, kbernoulli, normal, margulis

# make: rv definition
from ralgos.make import rv, nop

# infer: inference tools
from ralgos.infer.concentrate import chernoff, hoeffding
```

## Generators

```python
# Generators
g1 = kuniform(k=2,m=100)  # kwise uniform generator with k = 2, m = 100: samples from [0,100) 
g2 = uniform(m=80)    # independent unifrom generator with m = 80: samples from [0,80)
g3 = uuniform(m=90)    # universal unifrom generator with m = 90
g4 = normal(mean = 5, std=1)       # normal generator mu = 5, std = 1
g5 = bernoulli(p=0.6)    # bernoulli with p = 0.6
g6 = kbernoulli(k=2,p_=0.6)   # kwise bernoulli with k = 2, p = 0.6
g7 = margulis(5,100)  # margulis random walk generator with n = 5, m = 100
```

```python
print("Generators: ")
n = 10
generators = [g1,g2,g3,g4,g5,g6,g7]
for i in range(7):
    print(generators[i].type,": ",generators[i].sample(n))
```

## RV Construction

```python
# RV Definition
x1 = rv(g1, name="x1")
x2 = rv(g2, name="x2")
x3 = rv(g3, name="x3")
x4 = rv(g4, name="x4")

# functional form
# x5 = sum(sum(x1,x2, name="s1"),product(x3, x4, name="p1"),name="s2")

x5 = x1*x2 + x3*x4 + 2
x5.visualize()
```

```python
print(x5.sample(n))
```

## Infer

```python
# Normal RV
ga = normal(mean = 0, std=1)       # normal generator mu = 0, std = 1
xa = rv(ga,'xa')
print("xa: ",xa.sample(n))    # samping constructed rv


# Sum of Independet Bernoulli RVs
gb = bernoulli(p=0.6)    # bernoulli with p = 0.6
xb = nop(gb,"+",10,name="xb")
print("xb: ",xb.sample(n))    # samping constructed rv

# Sum of Independent Unifrom RVs
gc = uniform(m=100)    # independent unifrom generator with m = 10: samples from [0,10)
xc = nop(gc,"+",10,"xc")
print("xc: ",xc.sample(n))    # samping constructed rv
```

```python
print("\nChernoff bound on xa: ",chernoff(xa,3))
print("Hoeffding bound on xa: ",hoeffding(xa,3))


print("Chernoff bound on xb: ", chernoff(xb,9))
print("Hoeffding bound on xb: ", hoeffding(xb,9))


print("Chernoff bound on xc: ", chernoff(xc,750))
print("Hoeffding bound on xc: ", hoeffding(xc,750))
```
